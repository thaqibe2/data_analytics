# Analysis of contingency tables

## Libraries and settings

```python
# Libraries
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import chi2
from scipy.stats import chi2_contingency

# Ignore warnings
import warnings
warnings.filterwarnings('ignore')

# Get current working directory
print(os.getcwd())
```

## Import the apartment data

```python
# Define columns
columns = ['web-scraper-order',
            'address_raw',
            'lat',
            'lon',
            'bfs_number',
            'bfs_name',
            'rooms', 
            'area', 
            'luxurious', 
            'price', 
            'price_per_m2',
            'pop_dens',
            'frg_pct',
            'mean_taxable_income',
            'dist_supermarket']

# Read and select variables
df_orig = pd.read_csv("apartments_data_enriched_cleaned.csv", sep=";")[columns]

# Rename variable 'web-scraper-order' to 'apmt_id'
df_orig = df_orig.rename(columns={'web-scraper-order': 'apartment_id'})

# Remove missing values
df = df_orig.dropna()
df.head(5)

# Remove duplicates
df = df.drop_duplicates()
df.head(5)
```

### Create new categorical variable 'area_cat'

```python
# Create labels
labels = ['0 - 49', '50 - 74', '75 - 99', '100 - 149', '>= 150']

# Create new categorical variable
df["area_cat"] = pd.cut(df['area'], bins=[0, 50, 75, 100, 150, 1000], labels=labels)

# Check values
df[['area', 'area_cat']].head(10)
```

## Create a contingency table with the number of apartments per 'area_cat' versus 'price_cat'

```python
# Contingency table (numbers in the table are apartments)
tab = pd.pivot_table(df[['apartment_id', 'area_cat', 'price_cat']],
                     index=['area_cat'],
                     columns=['price_cat'], 
                     aggfunc='count')
tab
```

## Perform a Chi-square test on the contingency table
For details see: https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.chi2_contingency.html

```python
# 1-alpha
prob = 0.95

# Chi-squared test
stat, p, dof, expected = chi2_contingency(tab)

# Critical value
critical = chi2.ppf(prob, dof)

# Print results of the Chi-square test
print('Expected frequencies:', '\n', expected.round(4), '\n')
print('Critical value:', critical.round(4), '\n')
print('Test statistic:', stat.round(4), '\n')
print('Degrees of freedom:', dof, '\n')
print(f'p-value: {p:.4f}')
```

## Interpretation of the Chi-squared test result

<p>Because the test statistic is larger than the critical value and the p-value is smaller than our significance level apha=0.05, we can reject the null hypothesis, i.e. there is a statistically significant (at the 0.05 significance-level) association between the price-categories and the living-area categories in the contingence table.</p>

### Contingency table, spotted hyena example

```python
# The numpy array shows how many times each hyena responded (see study above)
table = np.array([[0,4],
                  [7,6],
                  [9,0],
                  [15,10]])

# Create a data frame from 'table'
df_hy = pd.DataFrame(table, columns=[['Female', 'Male']],
             index=['Fear', 'Greeting', 'Aggression', 'No Response'])
df_hy
```

### b) Perform Chi-squared test on the hyena contingency table

```python
# 1-alpha
prob = 0.95

# Chi-squared test on hyena data
stat, p, dof, expected = chi2_contingency(table)

# Critical value
critical = chi2.ppf(prob, dof)

# Print results of the Chi-square test
print('Expected frequencies:', '\n', expected.round(4), '\n')
print('Critical value:', critical.round(4), '\n')
print('Test statistic:', stat.round(4), '\n')
print('Degrees of freedom:', dof, '\n')
print(f'p-value: {p:.4f}')
```

### c) Python function to calculate expected frequencies

```python
def calculate_expected_frequencies(observed):
    """
    Calculate the expected frequencies for a contingency table.
    
    Parameters:
    observed: numpy array or pandas DataFrame with observed frequencies
    
    Returns:
    expected: numpy array with expected frequencies
    """
    # Convert to numpy array if DataFrame
    if isinstance(observed, pd.DataFrame):
        observed = observed.values
    
    # Calculate row totals
    row_totals = observed.sum(axis=1)
    
    # Calculate column totals
    col_totals = observed.sum(axis=0)
    
    # Calculate grand total
    grand_total = observed.sum()
    
    # Calculate expected frequencies
    # E_ij = (row_total_i * col_total_j) / grand_total
    expected = np.outer(row_totals, col_totals) / grand_total
    
    return expected

# Apply the function to the hyena data
expected_manual = calculate_expected_frequencies(table)
print('Expected frequencies (calculated manually):')
print(expected_manual.round(4))
```

### d) Compare calculated expected frequencies with chi2_contingency()

```python
# Get expected frequencies from chi2_contingency
_, _, _, expected_scipy = chi2_contingency(table)

print('Expected frequencies from chi2_contingency():')
print(expected_scipy.round(4))
print('\nExpected frequencies from our function:')
print(expected_manual.round(4))
print('\nAre the results equal?', np.allclose(expected_manual, expected_scipy))
```

### e) Extended function to calculate Chi-square test statistic

```python
def calculate_chi_square(observed):
    """
    Calculate the expected frequencies and Chi-square test statistic
    for a contingency table.
    
    Parameters:
    observed: numpy array or pandas DataFrame with observed frequencies
    
    Returns:
    expected: numpy array with expected frequencies
    chi_square_stat: float, the Chi-square test statistic
    """
    # Convert to numpy array if DataFrame
    if isinstance(observed, pd.DataFrame):
        observed = observed.values
    
    # Calculate row totals
    row_totals = observed.sum(axis=1)
    
    # Calculate column totals
    col_totals = observed.sum(axis=0)
    
    # Calculate grand total
    grand_total = observed.sum()
    
    # Calculate expected frequencies
    expected = np.outer(row_totals, col_totals) / grand_total
    
    # Calculate Chi-square test statistic
    # Chi-square = sum((O - E)^2 / E)
    chi_square_stat = np.sum((observed - expected) ** 2 / expected)
    
    return expected, chi_square_stat

# Apply the extended function to the hyena data
expected_ext, chi_square_manual = calculate_chi_square(table)
print(f'Chi-square test statistic (calculated manually): {chi_square_manual:.4f}')
```

### f) Compare calculated test statistic with chi2_contingency()

```python
# Get test statistic from chi2_contingency
stat_scipy, _, _, _ = chi2_contingency(table)

print(f'Chi-square test statistic from chi2_contingency(): {stat_scipy:.4f}')
print(f'Chi-square test statistic from our function: {chi_square_manual:.4f}')
print(f'\nAre the results equal? {np.isclose(chi_square_manual, stat_scipy)}')
```

### g) Conclusion on the original study results

Based on our Chi-squared test analysis of the hyena contingency table, the p-value is approximately 0.0065, which is less than the significance level alpha = 0.05. This means we reject the null hypothesis of independence, indicating there is a statistically significant association between the gender of the hyenas (Female/Male) and the type of response behavior (Fear, Greeting, Aggression, No Response). The results of the original study are correct - there is indeed a significant difference in behavioral responses between male and female spotted hyenas.
