# Exploratory data analysis (EDA) of supermarket data

This notebook performs exploratory data analysis on supermarket data in the canton of Zürich.

## Libraries and settings

```python
# Libraries
import os
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# seaborn graphics settings
sns.set(color_codes=True)

# Ignore warnings
import warnings
warnings.filterwarnings("ignore")

# Show current working directory
print(os.getcwd())
```

## Import supermarket data

```python
# Read supermarket data
df = pd.read_csv("supermarkets_data_enriched.csv")
df.head(5)
```

### Data shape and types

```python
print(f"Shape: {df.shape}")
print("\nData types:")
print(df.dtypes)
```

### Summary statistics

```python
df.describe()
```

## Count the number of supermarkets per brand

```python
# Count supermarkets per brand
brand_counts = df['brand'].value_counts()
print(brand_counts)
```

## Barchart with the number of supermarkets per brand

```python
# Create barchart of supermarkets per brand
brand_counts = df['brand'].value_counts().sort_values(ascending=True)

# Values for barchart
counts = list(brand_counts.values)
brands = list(brand_counts.index.values)
y_pos = np.arange(len(brands))

# Figure
fig, ax = plt.subplots(figsize=(10,6))
ax.barh(y_pos, counts, align='center', color='steelblue', alpha=0.8)
ax.set_yticks(y_pos, brands)
ax.set_xlabel('Number of supermarkets', fontsize=12)
ax.set_ylabel('Brand', fontsize=12)
ax.set_title('Number of supermarkets per brand', fontsize=14)

# Show graph
plt.tight_layout()
plt.show()
```

## Scatterplot-matrix of numeric variables

```python
# Create scatterplot matrix using PairGrid
sns.set(style="ticks", font_scale=0.8)
g = sns.PairGrid(df[['lat', 'lon', 'pop', 'pop_dens', 'frg_pct', 'emp']], height=1.5, aspect=1)
g.map_upper(sns.scatterplot, color='darkblue', s=10, alpha=0.5)
g.map_lower(sns.scatterplot, color='darkblue', s=10, alpha=0.5)
g.map_diag(plt.hist, color='steelblue')
plt.suptitle('Scatterplot matrix of numeric variables', y=1.02, fontsize=14)
plt.tight_layout()
plt.show()
```

## Plot supermarket locations by brand

```python
# Filter for Denner, Volg, and Landi
brands_to_plot = ['Denner', 'Volg', 'Landi']
df_subset = df[df['brand'].isin(brands_to_plot)]

# Create figure
fig, ax = plt.subplots(figsize=(10, 8))

# Define colors for each brand
colors = {'Denner': 'red', 'Volg': 'green', 'Landi': 'orange'}

# Plot each brand with different color
for brand in brands_to_plot:
    brand_data = df_subset[df_subset['brand'] == brand]
    ax.scatter(brand_data['lon'], brand_data['lat'], 
               c=colors[brand], label=brand, alpha=0.7, s=30)

# Set labels and title
ax.set_xlabel('Longitude', fontsize=12)
ax.set_ylabel('Latitude', fontsize=12)
ax.set_title('Supermarket Locations (Denner, Volg, Landi)', fontsize=14)
ax.legend(title='Brand', fontsize=10)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## Correlation heatmap

```python
# Set font size for plot
sns.set(font_scale=0.8)

# Create correlation matrix
plt.figure(figsize=(8,6))
numeric_cols = ['lat', 'lon', 'pop', 'pop_dens', 'frg_pct', 'emp']
corr = df[numeric_cols].corr().round(2)

# Plot heatmap
sns.heatmap(corr, cmap="coolwarm", annot=True)
plt.title('Correlation Heatmap of Numeric Variables', fontsize=14)
plt.tight_layout()
plt.show()
```
